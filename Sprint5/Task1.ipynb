<h1 style="color:DodgerBlue">HashSet<T></h1>



`HashSet<T>` в C# — это коллекция, которая хранит уникальные элементы без определенного порядка. Она полезна, когда нужно быстро проверять наличие элемента или добавлять только уникальные элементы. Давайте рассмотрим, как с ней работать на примере классов `Transport`, `Car`, и `Bike`.

### Создание классов Transport, Car, и Bike

Сначала определим базовый класс `Transport` и его наследники `Car` и `Bike`:

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public override bool Equals(object obj)
    {
        // Простая проверка равенства базируется на модели и производителе
        if (obj is Transport transport)
        {
            return Model == transport.Model && Manufacturer == transport.Manufacturer;
        }
        return false;
    }

    public override int GetHashCode()
    {
        // Создание хеш-кода на основе модели и производителя
        return HashCode.Combine(Model, Manufacturer);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Number of Doors: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Has Carrier: {HasCarrier}");
    }
}
```

### Работа с HashSet<T>

Теперь рассмотрим, как использовать `HashSet<T>` для хранения объектов `Transport`.

```csharp
using System;
using System.Collections.Generic;


// Создаем HashSet для уникальных объектов транспортных средств
HashSet<Transport> transportSet = new HashSet<Transport>();

// Добавляем объекты Car и Bike в HashSet
var car1 = new Car("Model S", "Tesla", 4);
var car2 = new Car("Model 3", "Tesla", 4);
 var bike1 = new Bike("Mountain Explorer", "Giant", true);

transportSet.Add(car1);
transportSet.Add(car2);
transportSet.Add(bike1);

// Попробуем добавить повторяющийся объект
var duplicateCar = new Car("Model S", "Tesla", 4);
bool isAdded = transportSet.Add(duplicateCar);

Console.WriteLine($"Was duplicate car added? {isAdded}"); // false

// Перебираем и выводим информацию обо всех уникальных транспортных средствах
foreach (var transport in transportSet)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

```

### Объяснения и комментарии:

- **Уникальность объектов**: `HashSet<T>` не допускает дублирования. Если вы пытаетесь добавить элемент, который уже присутствует (определяется методом `Equals` и `GetHashCode`), он не добавляется.

- **Методы `Equals` и `GetHashCode`**: Эти методы переопределены для корректной работы с `HashSet`. `Equals` определяет, когда два объекта считаются равными, а `GetHashCode` возвращает хеш-код, используемый для быстрого поиска элемента.

- **Быстрая проверка на наличие**: Благодаря хешированию, `HashSet` позволяет быстро проверять, существует ли элемент в коллекции (`O(1)` — амортизированное время).

- **Отсутствие порядка**: Элементы в `HashSet` не имеют определенного порядка, в отличие от `List<T>`.

`HashSet<T>` идеально подходит для сценариев, где требуется исключительно уникальность элементов и быстрое выполнение операций поиска.

````
using System;
using System.Collections.Generic;

// Базовый класс Animal
public abstract class Animal : IEquatable<Animal>
{
    public string Name { get; set; }
    public int Age { get; set; }
    public abstract string AnimalType { get; }
    
    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
    }
    
    public abstract void MakeSound();
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Тип: {AnimalType}, Имя: {Name}, Возраст: {Age} лет");
    }
    
    // Реализация IEquatable<Animal> для правильной работы HashSet
    public bool Equals(Animal other)
    {
        if (other is null) return false;
        return Name == other.Name && Age == other.Age && AnimalType == other.AnimalType;
    }
    
    public override bool Equals(object obj)
    {
        return Equals(obj as Animal);
    }
    
    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Age, AnimalType);
    }
    
    // Переопределяем операторы для сравнения
    public static bool operator ==(Animal left, Animal right)
    {
        if (left is null) return right is null;
        return left.Equals(right);
    }
    
    public static bool operator !=(Animal left, Animal right)
    {
        return !(left == right);
    }
}

// Производные классы
public class Dog : Animal
{
    public override string AnimalType => "Собака";
    public string Breed { get; set; }
    
    public Dog(string name, int age, string breed) : base(name, age)
    {
        Breed = breed;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} гавкает: Гав-гав!");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
    
    // Переопределяем Equals и GetHashCode с учетом специфических свойств
    public override bool Equals(object obj)
    {
        return Equals(obj as Dog);
    }
    
    public bool Equals(Dog other)
    {
        if (other is null) return false;
        return base.Equals(other) && Breed == other.Breed;
    }
    
    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), Breed);
    }
}

public class Cat : Animal
{
    public override string AnimalType => "Кот";
    public bool IsIndoor { get; set; }
    
    public Cat(string name, int age, bool isIndoor) : base(name, age)
    {
        IsIndoor = isIndoor;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Домашний: {(IsIndoor ? "Да" : "Нет")}");
    }
    
    public override bool Equals(object obj)
    {
        return Equals(obj as Cat);
    }
    
    public bool Equals(Cat other)
    {
        if (other is null) return false;
        return base.Equals(other) && IsIndoor == other.IsIndoor;
    }
    
    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), IsIndoor);
    }
}

public class Bird : Animal
{
    public override string AnimalType => "Птица";
    public double Wingspan { get; set; }
    
    public Bird(string name, int age, double wingspan) : base(name, age)
    {
        Wingspan = wingspan;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {Wingspan} см");
    }
    
    public override bool Equals(object obj)
    {
        return Equals(obj as Bird);
    }
    
    public bool Equals(Bird other)
    {
        if (other is null) return false;
        return base.Equals(other) && Wingspan == other.Wingspan;
    }
    
    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), Wingspan);
    }
}

public class Fish : Animal
{
    public override string AnimalType => "Рыба";
    public string WaterType { get; set; }
    
    public Fish(string name, int age, string waterType) : base(name, age)
    {
        WaterType = waterType;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} пускает пузыри: Буль-буль!");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}");
    }
    
    public override bool Equals(object obj)
    {
        return Equals(obj as Fish);
    }
    
    public bool Equals(Fish other)
    {
        if (other is null) return false;
        return base.Equals(other) && WaterType == other.WaterType;
    }
    
    public override int GetHashCode()
    {
        return HashCode.Combine(base.GetHashCode(), WaterType);
    }
}


        // Создание HashSet для животных
        HashSet<Animal> animalSet = new HashSet<Animal>();
        
        // Создание животных
        var dog1 = new Dog("Бобик", 3, "Овчарка");
        var dog2 = new Dog("Рекс", 4, "Дворняжка");
        var cat1 = new Cat("Мурзик", 2, true);
        var cat2 = new Cat("Барсик", 3, false);
        var bird1 = new Bird("Кеша", 1, 15.5);
        var fish1 = new Fish("Немо", 1, "Морская");
        
        // Добавление животных в HashSet
        Console.WriteLine("=== Добавление животных в HashSet ===");
        AddAnimalToSet(animalSet, dog1);
        AddAnimalToSet(animalSet, dog2);
        AddAnimalToSet(animalSet, cat1);
        AddAnimalToSet(animalSet, cat2);
        AddAnimalToSet(animalSet, bird1);
        AddAnimalToSet(animalSet, fish1);
        
        // Попытка добавить дубликат
        Console.WriteLine("\n=== Попытка добавить дубликат ===");
        var dog1Duplicate = new Dog("Бобик", 3, "Овчарка");
        AddAnimalToSet(animalSet, dog1Duplicate);
        
        // Вывод всех животных в HashSet
        Console.WriteLine("\n=== Все животные в HashSet ===");
        foreach (var animal in animalSet)
        {
            animal.DisplayInfo();
            animal.MakeSound();
            Console.WriteLine($"HashCode: {animal.GetHashCode()}");
            Console.WriteLine();
        }
        
        // Проверка наличия животных
        Console.WriteLine("=== Проверка наличия животных ===");
        var testDog = new Dog("Бобик", 3, "Овчарка");
        Console.WriteLine($"Содержит Бобика: {animalSet.Contains(testDog)}");
        
        var testCat = new Cat("Васька", 5, true); // Этого кота нет в множестве
        Console.WriteLine($"Содержит Ваську: {animalSet.Contains(testCat)}");
        
        // Удаление животного
        Console.WriteLine("\n=== Удаление животного ===");
        Console.WriteLine($"Количество животных до удаления: {animalSet.Count}");
        bool removed = animalSet.Remove(cat1);
        Console.WriteLine($"Удален Мурзик: {removed}");
        Console.WriteLine($"Количество животных после удаления: {animalSet.Count}");
        
        // Создание специализированных HashSet для каждого типа животных
        Console.WriteLine("\n=== Специализированные HashSet ===");
        HashSet<Dog> dogSet = new HashSet<Dog> { dog1, dog2 };
        HashSet<Cat> catSet = new HashSet<Cat> { cat1, cat2 };
        
        Console.WriteLine("Собаки:");
        foreach (var dog in dogSet)
        {
            dog.DisplayInfo();
        }
        
        Console.WriteLine("\nКошки:");
        foreach (var cat in catSet)
        {
            cat.DisplayInfo();
        }
        
        // Объединение множеств
        Console.WriteLine("\n=== Объединение множеств ===");
        var allAnimals = new HashSet<Animal>();
        allAnimals.UnionWith(animalSet);
        allAnimals.UnionWith(dogSet); // Добавляем собак (хотя они уже есть)
        allAnimals.UnionWith(catSet); // Добавляем котов
        
        Console.WriteLine($"Всего уникальных животных: {allAnimals.Count}");
   
    
    static void AddAnimalToSet(HashSet<Animal> set, Animal animal)
    {
        if (set.Add(animal))
        {
            Console.WriteLine($"Добавлено: {animal.AnimalType} {animal.Name}");
        }
        else
        {
            Console.WriteLine($"Дубликат! Не добавлено: {animal.AnimalType} {animal.Name}");
        }
    }
````
