<h1 style="color:DodgerBlue">Делегаты, лямбда-выражения и события<T></h1>



Делегаты, лямбда-выражения и события являются важными концепциями в C#, которые позволяют работать с методами как с объектами, писать краткий код и управлять реакциями на события. Рассмотрим их на примере гипотетических классов `Transport`, `Car`, и `Bike`.

### Делегаты

**Делегат** — это тип безопасного указателя на метод, похожий на функцию в C или C++. Делегаты в C# используются для передачи методов как параметров, а также для создания обратных вызовов.

**Пример:**

Создадим делегат, который может ссылаться на методы, выводящие информацию о транспортном средстве:

```csharp
public delegate void DisplayDelegate();

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}
```

Используем делегат:

```csharp

        Car car = new Car("Model S", "Tesla", 4);
        Bike bike = new Bike("Speedster", "Giant", true);

        // Присваиваем делегату метод DisplayInfo
        DisplayDelegate displayCar = car.DisplayInfo;
        DisplayDelegate displayBike = bike.DisplayInfo;

        // Вызов методов через делегаты
        displayCar();
        displayBike();




public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();

    }

{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Has Carrier: {HasCarrier}");
    }
}
```

### Лямбда-выражения

**Лямбда-выражения** позволяют вам писать анонимные методы в более краткой форме. Это часто используется в LINQ и асинхронном программировании.

**Пример:**

```csharp
using System;
using System.Collections.Generic;
using System.Linq;


 
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
};

// Используем лямбда-выражение для фильтрации транспортных средств
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var car in teslaCars)
{
    car.DisplayInfo();
}


```

### События

**События** используются для оповещения подписчиков о случившихся действиях, предоставляя более высокоуровневый способ взаимодействия между различными частями кода.

**Пример:**

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}



Transport transport = new Transport { Model = "Old Model", Manufacturer = "Generic" };
// Подписываем метод на событие
transport.TransportUpdated += OnTransportUpdated;
// Обновляем модель; это вызовет событие
transport.UpdateModel("New Model");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```

### Объяснения и комментарии:

- **Делегаты**: Тип данных, который можно использовать для хранения ссылок на методы. Полезно для обратных вызовов и передачи методов.

- **Лямбда-выражения**: Более краткая запись анонимных методов, часто используется в LINQ для фильтрации, проекций и агрегаций.

- **События**: Механизм уведомления объектов об изменениях состояния или других действиях. Позволяют подписываться на события в одной части программы и реакцию на них в другой.

Эти механизмы делают программу гибкой и более структурированной, обеспечивая четкое отделение логики и упрощая повторное использование кода.

````
using System;
using System.Collections.Generic;

// Делегаты
public delegate void AnimalSoundHandler(string sound);
public delegate void AnimalActionHandler(string action);
public delegate bool AnimalConditionHandler(Animal animal);

// Базовый класс Animal с событиями
public abstract class Animal
{
    // События
    public event AnimalSoundHandler OnMakeSound;
    public event AnimalActionHandler OnActionPerformed;
    public event EventHandler<AnimalEventArgs> OnAnimalEvent;

    public string Name { get; set; }
    public int Age { get; set; }
    public int Energy { get; protected set; } = 100;
    public abstract string AnimalType { get; }
    
    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
    }
    
    // Виртуальный метод для издания звука с вызовом события
    public virtual void MakeSound()
    {
        string sound = GetSound();
        OnMakeSound?.Invoke($"{Name} издает звук: {sound}");
        RaiseAnimalEvent("Sound", sound);
    }
    
    protected abstract string GetSound();
    
    // Метод для выполнения действия
    public virtual void PerformAction(string action)
    {
        OnActionPerformed?.Invoke($"{Name} выполняет: {action}");
        RaiseAnimalEvent("Action", action);
        Energy -= 10;
        
        if (Energy <= 0)
        {
            RaiseAnimalEvent("Tired", "Животное устало и нуждается в отдыхе");
        }
    }
    
    public void Rest()
    {
        Energy = Math.Min(100, Energy + 30);
        RaiseAnimalEvent("Rest", "Животное отдохнуло");
    }
    
    protected virtual void RaiseAnimalEvent(string eventType, string message)
    {
        OnAnimalEvent?.Invoke(this, new AnimalEventArgs(eventType, message, DateTime.Now));
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Тип: {AnimalType}, Имя: {Name}, Возраст: {Age}, Энергия: {Energy}%");
    }
}

// Класс для аргументов события
public class AnimalEventArgs : EventArgs
{
    public string EventType { get; }
    public string Message { get; }
    public DateTime Timestamp { get; }
    
    public AnimalEventArgs(string eventType, string message, DateTime timestamp)
    {
        EventType = eventType;
        Message = message;
        Timestamp = timestamp;
    }
}

// Производные классы
public class Dog : Animal
{
    public override string AnimalType => "Собака";
    public string Breed { get; set; }
    
    public Dog(string name, int age, string breed) : base(name, age)
    {
        Breed = breed;
    }
    
    protected override string GetSound() => "Гав-гав!";
    
    public void FetchBall()
    {
        PerformAction("приносит мячик");
    }
    
    public void GuardHouse()
    {
        PerformAction("охраняет дом");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
}

public class Cat : Animal
{
    public override string AnimalType => "Кот";
    public bool IsIndoor { get; set; }
    
    public Cat(string name, int age, bool isIndoor) : base(name, age)
    {
        IsIndoor = isIndoor;
    }
    
    protected override string GetSound() => "Мяу-мяу!";
    
    public void ClimbTree()
    {
        PerformAction("лазает по дереву");
    }
    
    public void HuntMouse()
    {
        PerformAction("охотится на мышь");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Домашний: {(IsIndoor ? "Да" : "Нет")}");
    }
}

public class Bird : Animal
{
    public override string AnimalType => "Птица";
    public double Wingspan { get; set; }
    
    public Bird(string name, int age, double wingspan) : base(name, age)
    {
        Wingspan = wingspan;
    }
    
    protected override string GetSound() => "Чик-чирик!";
    
    public void Fly()
    {
        PerformAction("летает в небе");
    }
    
    public void BuildNest()
    {
        PerformAction("строит гнездо");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {Wingspan} см");
    }
}

public class Fish : Animal
{
    public override string AnimalType => "Рыба";
    public string WaterType { get; set; }
    
    public Fish(string name, int age, string waterType) : base(name, age)
    {
        WaterType = waterType;
    }
    
    protected override string GetSound() => "Буль-буль!";
    
    public void Swim()
    {
        PerformAction("плавает в воде");
    }
    
    public void HideInCoral()
    {
        PerformAction("прячется в кораллах");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}");
    }
}

// Класс для управления животными с использованием делегатов
public class AnimalManager
{
    private List<Animal> animals = new List<Animal>();
    
    // Делегаты для фильтрации и действий
    public AnimalConditionHandler FilterCondition { get; set; }
    public Action<Animal> AnimalProcessor { get; set; }
    
    public void AddAnimal(Animal animal)
    {
        animals.Add(animal);
        
        // Подписываемся на события животного с помощью лямбда-выражений
        animal.OnMakeSound += sound => 
            Console.WriteLine($"[МЕНЕДЖЕР] Зарегистрирован звук: {sound}");
            
        animal.OnAnimalEvent += (sender, e) =>
        {
            var animalSender = sender as Animal;
            Console.WriteLine($"[МЕНЕДЖЕР] Событие от {animalSender.Name}: {e.EventType} - {e.Message} в {e.Timestamp:HH:mm:ss}");
        };
    }
    
    public void ProcessAnimals()
    {
        if (FilterCondition != null && AnimalProcessor != null)
        {
            var filteredAnimals = animals.FindAll(animal => FilterCondition(animal));
            filteredAnimals.ForEach(AnimalProcessor);
        }
    }
    
    public void PerformActionOnAll(Action<Animal> action)
    {
        animals.ForEach(action);
    }
}


        // Создание менеджера животных
        var manager = new AnimalManager();
        
        // Создание животных
        var dog = new Dog("Бобик", 3, "Овчарка");
        var cat = new Cat("Мурзик", 2, true);
        var bird = new Bird("Кеша", 1, 15.5);
        var fish = new Fish("Немо", 1, "Морская");
        
        // Подписка на события с помощью лямбда-выражений
        dog.OnMakeSound += sound => Console.WriteLine($"[СОБЫТИЕ] {sound}");
        dog.OnActionPerformed += action => Console.WriteLine($"[ДЕЙСТВИЕ] {action}");
        
        cat.OnMakeSound += sound => Console.WriteLine($"[СОБЫТИЕ] Кот говорит: {sound}");
        
        bird.OnAnimalEvent += (sender, e) => 
        {
            if (e.EventType == "Tired")
                Console.WriteLine($"[ВНИМАНИЕ] Птичке нужен отдых!");
        };
        
        // Добавление животных в менеджер
        manager.AddAnimal(dog);
        manager.AddAnimal(cat);
        manager.AddAnimal(bird);
        manager.AddAnimal(fish);
        
        Console.WriteLine("=== ДЕМОНСТРАЦИЯ СОБЫТИЙ ===");
        dog.MakeSound();
        cat.MakeSound();
        bird.MakeSound();
        
        Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ ДЕЙСТВИЙ ===");
        dog.FetchBall();
        cat.ClimbTree();
        bird.Fly();
        fish.Swim();
        
        // Использование делегатов для фильтрации
        Console.WriteLine("\n=== ФИЛЬТРАЦИЯ С ИСПОЛЬЗОВАНИЕМ ДЕЛЕГАТОВ ===");
        
        // Фильтр для молодых животных (лямбда-выражение)
        manager.FilterCondition = animal => animal.Age < 3;
        manager.AnimalProcessor = animal => 
        {
            Console.WriteLine($"Молодое животное: {animal.Name} ({animal.Age} лет)");
        };
        manager.ProcessAnimals();
        
        // Фильтр для животных с низкой энергией
        Console.WriteLine("\n=== ЖИВОТНЫЕ С НИЗКОЙ ЭНЕРГИЕЙ ===");
        manager.FilterCondition = animal => animal.Energy < 80;
        manager.AnimalProcessor = animal => 
        {
            Console.WriteLine($"{animal.Name} устал (энергия: {animal.Energy}%)");
            animal.Rest(); // Даем отдохнуть
        };
        manager.ProcessAnimals();
        
        // Использование Action с лямбда-выражениями
        Console.WriteLine("\n=== ВЫПОЛНЕНИЕ ДЕЙСТВИЙ НАД ВСЕМИ ЖИВОТНЫМИ ===");
        manager.PerformActionOnAll(animal => 
        {
            animal.MakeSound();
            animal.DisplayInfo();
            Console.WriteLine();
        });
        
        // Комбинирование делегатов
        Console.WriteLine("=== КОМБИНИРОВАНИЕ ДЕЛЕГАТОВ ===");
        AnimalConditionHandler youngFilter = a => a.Age < 2;
        AnimalConditionHandler highEnergyFilter = a => a.Energy > 50;
        
        // Комбинируем фильтры
        AnimalConditionHandler combinedFilter = a => youngFilter(a) && highEnergyFilter(a);
        
        manager.FilterCondition = combinedFilter;
        manager.AnimalProcessor = animal =>
        {
            Console.WriteLine($"Молодое и энергичное: {animal.Name}");
        };
        manager.ProcessAnimals();
        
        // Демонстрация Func делегата
        Console.WriteLine("\n=== ИСПОЛЬЗОВАНИЕ FUNC ДЕЛЕГАТА ===");
        Func<Animal, string> animalDescription = a => 
            $"{a.AnimalType} {a.Name} (возраст: {a.Age}, энергия: {a.Energy}%)";
        
        manager.PerformActionOnAll(animal => 
        {
            string description = animalDescription(animal);
            Console.WriteLine(description);
        });
        
        // Групповые действия с событиями
        Console.WriteLine("\n=== ГРУППОВЫЕ ДЕЙСТВИЯ ===");
        var animals = new List<Animal> { dog, cat, bird, fish };
        
        // Подписка на групповое событие
        foreach (var animal in animals)
        {
            animal.OnMakeSound += sound => 
                Console.WriteLine($"[ГРУППА] Получен звук от животного");
        }
        
        // Вызов звуков у всех животных
        animals.ForEach(a => a.MakeSound());
````
