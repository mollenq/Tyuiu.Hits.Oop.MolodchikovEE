<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Supplier в C#, который будет представлять информацию о поставщиках товаров или услуг. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Supplier:


Атрибуты: ID поставщика (SupplierId), Название компании (CompanyName), Тип продукции (ProductType).


Методы:


GetCompanyInfo(): метод для получения информации о компании.


ProvideQuote(): метод для предоставления котировки на товары или услуги.


SubmitOrder(): метод для отправки заказа поставщику.
Требования к производным классам:

1. Производитель (Manufacturer): Должен содержать дополнительные атрибуты, такие как Год основания (FoundedYear). Метод ProvideQuote() должен быть переопределен для включения информации о годе основания компании в котировку.
2. Ритейлер (Retailer): Должен содержать дополнительные атрибуты, такие как Расположение магазина (StoreLocation). Метод SubmitOrder() должен быть переопределен для добавления информации о расположении магазина при отправке заказа.
3. Импортер (Importer) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Страна происхождения товара (OriginCountry). Метод GetCompanyInfo() должен быть переопределен для отображения страны происхождения товара вместе с остальной информацией о компании.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) создайте явную реализации интерфейса и управление зависимостями

```
using System;

// Интерфейс для логирования
public interface ILogger
{
    void Log(string message);
}

// Простая реализация логгера
public class ConsoleLogger : ILogger
{
    public void Log(string message)
    {
        Console.WriteLine($"[LOG]: {message}");
    }
}

// Интерфейс для поставщиков (явная реализация)
public interface ISupplierActions
{
    void MakeOffer();
    void DeliverGoods();
}

// Базовый класс Supplier
public class Supplier : ISupplierActions
{
    protected readonly ILogger _logger; // зависимость через DI

    public int SupplierId { get; set; }
    public string CompanyName { get; set; }
    public string ProductType { get; set; }
    public string ContactPerson { get; set; }   // новый атрибут
    public string Email { get; set; }           // новый атрибут
    public double Rating { get; set; }          // новый атрибут

    public Supplier(int supplierId, string companyName, string productType, string contactPerson, string email, double rating, ILogger logger)
    {
        SupplierId = supplierId;
        CompanyName = companyName;
        ProductType = productType;
        ContactPerson = contactPerson;
        Email = email;
        Rating = rating;
        _logger = logger;
    }

    public virtual string GetCompanyInfo()
    {
        return $"ID: {SupplierId}, Компания: {CompanyName}, Контакт: {ContactPerson}, Email: {Email}, Рейтинг: {Rating}, Продукция: {ProductType}";
    }

    public virtual string ProvideQuote()
    {
        return $"Компания {CompanyName} предоставляет котировку на продукцию типа {ProductType}.";
    }

    public virtual string SubmitOrder()
    {
        return $"Заказ отправлен в компанию {CompanyName}.";
    }

    public void UpdateRating(double newRating)
    {
        Rating = newRating;
        _logger.Log($"Рейтинг компании {CompanyName} обновлён до {Rating}");
    }

    public void SendEmail(string message)
    {
        _logger.Log($"Отправлено письмо компании {CompanyName} ({Email}): {message}");
    }

    // Явная реализация интерфейса
    void ISupplierActions.MakeOffer()
    {
        _logger.Log($"{CompanyName} предлагает скидку на {ProductType}!");
    }

    void ISupplierActions.DeliverGoods()
    {
        _logger.Log($"{CompanyName} доставила продукцию {ProductType}.");
    }
}

// Производный класс Manufacturer
public class Manufacturer : Supplier
{
    public int FoundedYear { get; set; }
    public int EmployeesCount { get; set; }    // новый атрибут
    public string FactoryLocation { get; set; } // новый атрибут
    public bool HasISO { get; set; }           // новый атрибут

    public Manufacturer(int supplierId, string companyName, string productType, string contactPerson, string email, double rating,
                        int foundedYear, int employeesCount, string factoryLocation, bool hasISO, ILogger logger)
        : base(supplierId, companyName, productType, contactPerson, email, rating, logger)
    {
        FoundedYear = foundedYear;
        EmployeesCount = employeesCount;
        FactoryLocation = factoryLocation;
        HasISO = hasISO;
    }

    public override string ProvideQuote()
    {
        return $"Компания {CompanyName} (основана в {FoundedYear}) предоставляет котировку на {ProductType}.";
    }

    public void ProduceGoods()
    {
        _logger.Log($"{CompanyName} начала производство {ProductType} на фабрике в {FactoryLocation}.");
    }
}

// Производный класс Retailer
public class Retailer : Supplier
{
    public string StoreLocation { get; set; }
    public int BranchesCount { get; set; }     // новый атрибут
    public bool HasOnlineStore { get; set; }   // новый атрибут
    public string WorkingHours { get; set; }   // новый атрибут

    public Retailer(int supplierId, string companyName, string productType, string contactPerson, string email, double rating,
                    string storeLocation, int branchesCount, bool hasOnlineStore, string workingHours, ILogger logger)
        : base(supplierId, companyName, productType, contactPerson, email, rating, logger)
    {
        StoreLocation = storeLocation;
        BranchesCount = branchesCount;
        HasOnlineStore = hasOnlineStore;
        WorkingHours = workingHours;
    }

    public override string SubmitOrder()
    {
        return $"Заказ отправлен в компанию {CompanyName}, магазин в {StoreLocation}.";
    }

    public void OpenNewBranch()
    {
        BranchesCount++;
        _logger.Log($"Компания {CompanyName} открыла новый филиал. Теперь их {BranchesCount}.");
    }
}

// Производный класс Importer
public class Importer : Supplier
{
    public string OriginCountry { get; set; }
    public double ImportTax { get; set; }     // новый атрибут
    public string CustomsCode { get; set; }   // новый атрибут
    public bool Certified { get; set; }       // новый атрибут

    public Importer(int supplierId, string companyName, string productType, string contactPerson, string email, double rating,
                    string originCountry, double importTax, string customsCode, bool certified, ILogger logger)
        : base(supplierId, companyName, productType, contactPerson, email, rating, logger)
    {
        OriginCountry = originCountry;
        ImportTax = importTax;
        CustomsCode = customsCode;
        Certified = certified;
    }

    public override string GetCompanyInfo()
    {
        return base.GetCompanyInfo() + $", Страна происхождения: {OriginCountry}, Сертифицирован: {Certified}";
    }

    public void ClearCustoms()
    {
        _logger.Log($"{CompanyName} прошла таможню по коду {CustomsCode}.");
    }
}

// Тест программы

ILogger logger = new ConsoleLogger();

ISupplierActions manufacturer = new Manufacturer(1, "TechProd", "Электроника", "Иванов И.И.", "info@techprod.ru", 4.8,
    1995, 120, "Москва", true, logger);

ISupplierActions retailer = new Retailer(2, "ShopWorld", "Одежда", "Петров П.П.", "shop@world.ru", 4.5,
    "Москва", 15, true, "9:00–21:00", logger);

ISupplierActions importer = new Importer(3, "GlobalTrade", "Фрукты", "Гарсия Л.", "trade@global.com", 4.6,
    "Испания", 12.5, "FT-9821", true, logger);

// Вызов методов через интерфейс (явная реализация)
manufacturer.MakeOffer();
manufacturer.DeliverGoods();

retailer.MakeOffer();
retailer.DeliverGoods();

importer.MakeOffer();
importer.DeliverGoods();

Console.WriteLine();

// Приведение к базовому классу
Supplier sup = (Supplier)manufacturer;
Console.WriteLine(sup.GetCompanyInfo());
sup.UpdateRating(4.9);
sup.SendEmail("Спасибо за сотрудничество!");

```
