<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Supplier в C#, который будет представлять информацию о поставщиках товаров или услуг. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Supplier:


Атрибуты: ID поставщика (SupplierId), Название компании (CompanyName), Тип продукции (ProductType).


Методы:


GetCompanyInfo(): метод для получения информации о компании.


ProvideQuote(): метод для предоставления котировки на товары или услуги.


SubmitOrder(): метод для отправки заказа поставщику.
Требования к производным классам:

1. Производитель (Manufacturer): Должен содержать дополнительные атрибуты, такие как Год основания (FoundedYear). Метод ProvideQuote() должен быть переопределен для включения информации о годе основания компании в котировку.
2. Ритейлер (Retailer): Должен содержать дополнительные атрибуты, такие как Расположение магазина (StoreLocation). Метод SubmitOrder() должен быть переопределен для добавления информации о расположении магазина при отправке заказа.
3. Импортер (Importer) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Страна происхождения товара (OriginCountry). Метод GetCompanyInfo() должен быть переопределен для отображения страны происхождения товара вместе с остальной информацией о компании.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

// ---------- Базовый класс ----------
public class Supplier
{
    public int SupplierId { get; set; }
    public string CompanyName { get; set; }
    public string ProductType { get; set; }
    public string ContactEmail { get; set; }
    public string Phone { get; set; }

    public Supplier(int supplierId, string companyName, string productType, string contactEmail, string phone)
    {
        SupplierId = supplierId;     // через set
        CompanyName = companyName;   // через set
        ProductType = productType;   // через set
        ContactEmail = contactEmail; // через set
        Phone = phone;               // через set
    }

    public virtual string GetCompanyInfo()
    {
        return $"ID: {SupplierId}\nКомпания: {CompanyName}\nТип продукции: {ProductType}\nEmail: {ContactEmail}\nТелефон: {Phone}";
    }

    public virtual string ProvideQuote()
    {
        return $"Компания {CompanyName} предоставляет котировку на продукцию типа {ProductType}.";
    }

    public virtual string SubmitOrder()
    {
        return $"Заказ отправлен в компанию {CompanyName}.";
    }

    public virtual string Collaborate(Supplier partner)
    {
        return $"{CompanyName} сотрудничает с {partner.CompanyName} в сфере {ProductType}.";
    }

    public virtual string SendEmail(string message)
    {
        return $"Email отправлен в {ContactEmail}: {message}";
    }
}

// ---------- Интерфейсы для множественного наследования ----------
public interface ILogistics
{
    string DeliverGoods(string destination);
}

public interface IMarketing
{
    string AdvertiseProduct();
}

// ---------- Сложное наследование ----------
public class Manufacturer : Supplier, ILogistics, IMarketing
{
    public int FoundedYear { get; set; }
    public int NumberOfEmployees { get; set; }
    public double AnnualRevenue { get; set; }

    public Manufacturer(int supplierId, string companyName, string productType, string contactEmail, string phone,
                        int foundedYear, int employees, double revenue)
        : base(supplierId, companyName, productType, contactEmail, phone)
    {
        FoundedYear = foundedYear;
        NumberOfEmployees = employees;
        AnnualRevenue = revenue;
    }

    public override string ProvideQuote()
    {
        return $"Компания {CompanyName} (основана в {FoundedYear}) с оборотом {AnnualRevenue}$ предоставляет котировку на {ProductType}.";
    }

    public string DeliverGoods(string destination)
    {
        return $"Компания {CompanyName} доставляет товары в {destination}.";
    }

    public string AdvertiseProduct()
    {
        return $"Компания {CompanyName} запускает рекламу продукции {ProductType}.";
    }
}

// ---------- Простое наследование ----------
public class Retailer : Supplier
{
    public string StoreLocation { get; set; }
    public int NumberOfBranches { get; set; }
    public bool OnlineStore { get; set; }

    public Retailer(int supplierId, string companyName, string productType, string contactEmail, string phone,
                    string storeLocation, int branches, bool onlineStore)
        : base(supplierId, companyName, productType, contactEmail, phone)
    {
        StoreLocation = storeLocation;
        NumberOfBranches = branches;
        OnlineStore = onlineStore;
    }

    public override string SubmitOrder()
    {
        return $"Заказ отправлен в магазин {CompanyName}, расположенный в {StoreLocation}.";
    }

    public string LaunchSale()
    {
        return $"Магазин {CompanyName} запускает распродажу. Онлайн: {OnlineStore}";
    }
}

// ---------- Простое наследование ----------
public class Importer : Supplier
{
    public string OriginCountry { get; set; }
    public double CustomsDuty { get; set; }
    public string ImportLicense { get; set; }

    public Importer(int supplierId, string companyName, string productType, string contactEmail, string phone,
                    string originCountry, double customsDuty, string license)
        : base(supplierId, companyName, productType, contactEmail, phone)
    {
        OriginCountry = originCountry;
        CustomsDuty = customsDuty;
        ImportLicense = license;
    }

    public override string GetCompanyInfo()
    {
        return $"ID: {SupplierId}\nКомпания: {CompanyName}\nТип продукции: {ProductType}\nEmail: {ContactEmail}\nТелефон: {Phone}\nСтрана происхождения: {OriginCountry}\nЛицензия: {ImportLicense}";
    }

    public string CalculateFinalPrice(double basePrice)
    {
        double finalPrice = basePrice + basePrice * CustomsDuty;
        return $"Финальная цена после пошлины: {finalPrice}$";
    }
}

// ---------- Тест программы ----------


Supplier manufacturer = new Manufacturer(1, "TechProd", "Электроника", "info@techprod.com", "+123456", 1995, 500, 12000000);
Supplier retailer = new Retailer(2, "ShopWorld", "Одежда", "shop@world.com", "+987654", "Москва", 12, true);
Supplier importer = new Importer(3, "GlobalTrade", "Фрукты", "import@trade.com", "+567890", "Испания", 0.15, "LIC-2025");

Console.WriteLine("=== Manufacturer ===");
Console.WriteLine(manufacturer.GetCompanyInfo());
Console.WriteLine(manufacturer.ProvideQuote());
Console.WriteLine(((Manufacturer)manufacturer).DeliverGoods("Берлин"));
Console.WriteLine(((Manufacturer)manufacturer).AdvertiseProduct());
Console.WriteLine();

Console.WriteLine("=== Retailer ===");
Console.WriteLine(retailer.GetCompanyInfo());
Console.WriteLine(retailer.ProvideQuote());
Console.WriteLine(retailer.SubmitOrder());
Console.WriteLine(((Retailer)retailer).LaunchSale());
Console.WriteLine();

Console.WriteLine("=== Importer ===");
Console.WriteLine(importer.GetCompanyInfo());
Console.WriteLine(importer.ProvideQuote());
Console.WriteLine(importer.SubmitOrder());
Console.WriteLine(((Importer)importer).CalculateFinalPrice(100));
Console.WriteLine();

// Взаимодействие объектов
Console.WriteLine("=== Interaction ===");
Console.WriteLine(manufacturer.Collaborate(importer));
Console.WriteLine(importer.Collaborate(retailer));
Console.WriteLine(retailer.Collaborate(manufacturer));


```
