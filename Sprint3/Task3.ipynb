<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Supplier в C#, который будет представлять информацию о поставщиках товаров или услуг. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Supplier:


Атрибуты: ID поставщика (SupplierId), Название компании (CompanyName), Тип продукции (ProductType).


Методы:


GetCompanyInfo(): метод для получения информации о компании.


ProvideQuote(): метод для предоставления котировки на товары или услуги.


SubmitOrder(): метод для отправки заказа поставщику.
Требования к производным классам:

1. Производитель (Manufacturer): Должен содержать дополнительные атрибуты, такие как Год основания (FoundedYear). Метод ProvideQuote() должен быть переопределен для включения информации о годе основания компании в котировку.
2. Ритейлер (Retailer): Должен содержать дополнительные атрибуты, такие как Расположение магазина (StoreLocation). Метод SubmitOrder() должен быть переопределен для добавления информации о расположении магазина при отправке заказа.
3. Импортер (Importer) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Страна происхождения товара (OriginCountry). Метод GetCompanyInfo() должен быть переопределен для отображения страны происхождения товара вместе с остальной информацией о компании.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```

    // Обобщённый базовый класс
    public class Supplier<T>
    {
        public T SupplierId { get; set; }
        public string CompanyName { get; set; }
        public string ProductType { get; set; }
        public string Country { get; set; }
        public double Rating { get; set; }
        public int EmployeeCount { get; set; }

        public Supplier(T supplierId, string companyName, string productType, string country, double rating, int employeeCount)
        {
            SupplierId = supplierId;
            CompanyName = companyName;
            ProductType = productType;
            Country = country;
            Rating = rating;
            EmployeeCount = employeeCount;
        }

        public virtual string GetCompanyInfo()
        {
            return $"ID: {SupplierId}\nКомпания: {CompanyName}\nТип: {ProductType}\nСтрана: {Country}\nРейтинг: {Rating}\nСотрудников: {EmployeeCount}";
        }

        public virtual string ProvideQuote()
        {
            return $"Компания {CompanyName} предоставляет котировку на продукцию {ProductType}.";
        }

        public virtual string SubmitOrder()
        {
            return $"Заказ отправлен в компанию {CompanyName}.";
        }

        // Перегрузка метода
        public string SubmitOrder(string orderNumber)
        {
            return $"Заказ №{orderNumber} отправлен в компанию {CompanyName}.";
        }

        // Новый метод
        public void IncreaseRating(double value)
        {
            Rating += value;
            Console.WriteLine($"{CompanyName}: рейтинг увеличен до {Rating}");
        }
    }

    // Производный класс — Производитель
    public class Manufacturer<T> : Supplier<T>
    {
        public int FoundedYear { get; set; }
        public int FactoryCount { get; set; }
        public string CEO { get; set; }

        public Manufacturer(T supplierId, string companyName, string productType, string country, double rating, int employeeCount, int foundedYear, int factoryCount, string ceo)
            : base(supplierId, companyName, productType, country, rating, employeeCount)
        {
            FoundedYear = foundedYear;
            FactoryCount = factoryCount;
            CEO = ceo;
        }

        public override string ProvideQuote()
        {
            return $"Компания {CompanyName} (основана в {FoundedYear}, CEO: {CEO}) предоставляет котировку на {ProductType}.";
        }

        public string ProvideQuote(bool includeDiscount)
        {
            return includeDiscount
                ? $"{CompanyName} предоставляет скидочную котировку на {ProductType}!"
                : ProvideQuote();
        }
    }

    // Производный класс — Розничный продавец
    public class Retailer<T> : Supplier<T>
    {
        public string StoreLocation { get; set; }
        public int BranchCount { get; set; }
        public bool OnlineStore { get; set; }

        public Retailer(T supplierId, string companyName, string productType, string country, double rating, int employeeCount, string storeLocation, int branchCount, bool onlineStore)
            : base(supplierId, companyName, productType, country, rating, employeeCount)
        {
            StoreLocation = storeLocation;
            BranchCount = branchCount;
            OnlineStore = onlineStore;
        }

        public override string SubmitOrder()
        {
            return $"Заказ отправлен в магазин {CompanyName}, расположенный в {StoreLocation}.";
        }

        public string GetRetailInfo()
        {
            return $"Магазин: {CompanyName}\nФилиалов: {BranchCount}\nОнлайн-магазин: {(OnlineStore ? "Да" : "Нет")}";
        }
    }

    // Производный класс — Импортер
    public class Importer<T> : Supplier<T>
    {
        public string OriginCountry { get; set; }
        public double ImportTax { get; set; }
        public bool Certified { get; set; }

        public Importer(T supplierId, string companyName, string productType, string country, double rating, int employeeCount, string originCountry, double importTax, bool certified)
            : base(supplierId, companyName, productType, country, rating, employeeCount)
        {
            OriginCountry = originCountry;
            ImportTax = importTax;
            Certified = certified;
        }

        public override string GetCompanyInfo()
        {
            return $"ID: {SupplierId}\nКомпания: {CompanyName}\nПродукция: {ProductType}\nСтрана: {Country}\nСтрана происхождения: {OriginCountry}\nНалог: {ImportTax}%\nСертификация: {(Certified ? "Да" : "Нет")}";
        }

        public void CalculateTotalCost(double basePrice)
        {
            double total = basePrice + (basePrice * ImportTax / 100);
            Console.WriteLine($"{CompanyName}:\nБазовая цена: {basePrice} у.е.\nНалог: {ImportTax}%\nИтоговая стоимость: {total} у.е.");
        }
    }

    // Generic-класс для управления поставщиками
    public class SupplierManager<T>
    {
        private List<Supplier<T>> suppliers = new List<Supplier<T>>();

        public void AddSupplier(Supplier<T> supplier)
        {
            suppliers.Add(supplier);
        }

        public void ShowAll()
        {
            foreach (var s in suppliers)
            {
                Console.WriteLine("==================================");
                Console.WriteLine(s.GetCompanyInfo());
                Console.WriteLine(s.ProvideQuote());
                Console.WriteLine(s.SubmitOrder());
                Console.WriteLine("==================================\n");
            }
        }
    }


var manufacturer = new Manufacturer<int>(1, "TechProd", "Электроника", "Германия", 4.8, 350, 1995, 5, "Михаэль Шульц");
var retailer = new Retailer<int>(2, "ShopWorld", "Одежда", "Россия", 4.2, 120, "Москва", 12, true);
var importer = new Importer<int>(3, "GlobalTrade", "Фрукты", "Испания", 4.5, 200, "Чили", 10.5, true);

Console.WriteLine("=== Производитель ===");
Console.WriteLine(manufacturer.GetCompanyInfo());
Console.WriteLine(manufacturer.ProvideQuote(true));
Console.WriteLine(manufacturer.SubmitOrder("A-1001"));
manufacturer.IncreaseRating(0.1);
Console.WriteLine();

Console.WriteLine("=== Розничный продавец ===");
Console.WriteLine(retailer.GetCompanyInfo());
Console.WriteLine(retailer.ProvideQuote());
Console.WriteLine(retailer.SubmitOrder());
Console.WriteLine(retailer.GetRetailInfo());
Console.WriteLine();

Console.WriteLine("=== Импортер ===");
Console.WriteLine(importer.GetCompanyInfo());
importer.CalculateTotalCost(1000);
Console.WriteLine(importer.ProvideQuote());
Console.WriteLine(importer.SubmitOrder());
Console.WriteLine();

var manager = new SupplierManager<int>();
manager.AddSupplier(manufacturer);
manager.AddSupplier(retailer);
manager.AddSupplier(importer);

Console.WriteLine("=== Все поставщики ===\n");
manager.ShowAll();

```
